```python
import g2_lib as g2

network = g2.Network([])

bandwidth = 900

## Creation of a network
network = network.add_link(g2.g2_lib_compiled_bindings.Link('1-2', bandwidth))
network = network.add_link(g2.g2_lib_compiled_bindings.Link('2-3', bandwidth))
network = network.add_link(g2.g2_lib_compiled_bindings.Link('3-4', bandwidth))
network = network.add_link(g2.g2_lib_compiled_bindings.Link('4-1', bandwidth))

# Add another flowgroup for a long path in the opposite direction (1-8-7-6-5)
path_12 = ["1-2"]
network = network.add_flowgroup(path_12, "ring_1-2", 1)
path_23 = ["2-3"]
network = network.add_flowgroup(path_23, "ring_2-3", 1)
path_34 = ["3-4"]
network = network.add_flowgroup(path_34, "ring_3-4", 1)
path_41 = ["4-1", "3-4", "2-3", "1-2"]
network = network.add_flowgroup(path_41, "ring_4-1", 1)

for f in network.flow_names():
    print(f"Flowgroup: {f}, Flow rate: {network.get_flow_rate(f)}")
```
