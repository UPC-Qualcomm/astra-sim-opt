```python
# --- Dragonfly Configuration ---
# G=4 groups, A=4 switches/group, h=2 inter-group links, 1 host/switch
# Total Hosts = 4 * 4 * 1 = 16
# Bandwidth levels: host_switch, intra_group, inter_group
df_bw = {'host_switch': 100, 'intra_group': 50, 'inter_group': 25}
df_config = {'G': 4, 'A': 4, 'h': 2, 'concentration': 1, 'bandwidth_config': df_bw, 'bw_unit': 'GB/s'}
generate_topology_files(topology="Dragonfly", paths_mode="ECMP", config=df_config)

# --- Jellyfish Configuration ---
# 16 switches, degree 4, 1 host/switch
# Total Hosts = 16 * 1 = 16
# Bandwidth levels: host_switch, switch_switch
jf_bw = {'host_switch': 100, 'switch_switch': 50}
jf_config = {'num_switches': 16, 'degree': 4, 'num_hosts_per_switch': 1, 'bandwidth_config': jf_bw, 'bw_unit': 'GB/s'}
generate_topology_files(topology="Jellyfish", paths_mode="Uniform", config=jf_config)

# --- Folded Clos (Fat-Tree) Configuration ---
# K=4 (4-port switches)
# Total Hosts = (4^3) / 4 = 16
# Bandwidth levels: host_edge, edge_agg, agg_core
fc_bw = {'host_edge': 100, 'edge_agg': 100, 'agg_core': 100}
fc_config = {'K': 4, 'bandwidth_config': fc_bw, 'bw_unit': 'MiB/s'}
generate_topology_files(topology="FoldedClos", paths_mode="ECMP", config=fc_config)
```
