```python
fc = FoldedClos(6, 1, 1)
fc.DesignFullTopology()
links, str_builder = fc.LinksToG2ConfFile()
print(fc.links)
fc.GenerateECMPFlowDict(fc.adjacency_matrix)
paths = fc.paths
print("Paths:", paths)
paths = fc.GenerateUniformRouting()
for src, dests in paths.items():
    for dest, path in dests.items():
        # Check if the path is a single list (i.e., not already a list of lists)
        if isinstance(path, list) and (not path or isinstance(path[0], str)):
            paths[src][dest] = [path]
print("Paths:", paths)
write_g2_topology_files(links, paths, base_filename="topology_g2_54NPUs", bandwidth=2)
write_ns3_topology_file(links, paths, filename="topology_ns3_54NPUs", bandwidth="0.002GiB/s")
```
