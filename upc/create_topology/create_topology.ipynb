```python
import random
k = 4
fc = FoldedClos(k, 1, 1)
fc.DesignFullTopology()
links, str_builder = fc.LinksToG2ConfFile()
print(fc.links)

##### GENERATE ECMP ROUTING PATHS #######
fc.GenerateECMPFlowDict(fc.adjacency_matrix)
paths = fc.paths
##########################################

##### GENERATE UNIFORM ROUTING PATHS #####
# print("Paths:", paths)
# paths = fc.GenerateUniformRouting()
##########################################
hosts = 4**3 / 4
topos = 5


for topo in range(topos):
    paths_new = {}
    for src, dests in paths.items():
        paths_new[src] = {}
        for dest, path_list in dests.items():
            paths_new[src][dest] = [random.choice(path_list)]
    print("Paths:", paths)
    write_g2_topology_files(links, paths_new, base_filename=f"G2_FoldedClos_{hosts}_topology{topo}", bandwidth=2)
    write_ns3_topology_file(links, paths_new, filename=f"ns3_FoldedClos_{hosts}_topology{topo}", bandwidth="2MB/s")
```
