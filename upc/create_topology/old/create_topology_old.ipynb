```python
from create_topology import CustomizedDragonfly, Jellyfish, FoldedClos
from utils import write_ns3_topology_file, write_g2_topology_files
import random
```

```python
cd = CustomizedDragonfly(4, 4, 2, 2)
cd.DesignFullTopology()
links = cd.LinksToG2ConfFile()
print("Links:", links)
paths = cd.GenerateUniformRouting()
print("Paths:", paths)

##### GENERATE ECMP ROUTING PATHS #######
cd.GenerateECMPFlowDict(cd.adjacency_matrix)
paths = cd.paths
##########################################
for host1 in range(cd.total_num_hosts):
    paths[f'h{host1+1}'] = {}
    for host2 in range(cd.total_num_hosts):
        if host1 != host2:
            s_host1 = (host1)//cd.concentration_factor
            s_host2 = (host2)//cd.concentration_factor
            if s_host1 == s_host2:
                paths[f'h{host1+1}'][f'h{host2+1}'] = [[f'h{host1+1}'] + [f's{s_host1+1}'] + [f'h{host2+1}']]
            else:   
                # paths[f's{s_host1+1}'][f's{s_host2+1}'] is a list of lists, so we need to iterate through each path
                paths[f'h{host1+1}'][f'h{host2+1}'] = [[f'h{host1+1}'] + path + [f'h{host2+1}'] for path in paths[f's{s_host1+1}'][f's{s_host2+1}']]

hosts = cd.num_groups * cd.num_switches * cd.concentration_factor
topos = 5

for topo in range(topos):
    paths_new = {}
    for src, dests in paths.items():
        paths_new[src] = {}
        for dest, path_list in dests.items():
            paths_new[src][dest] = [random.choice(path_list)]
    print("Paths:", paths)
    write_g2_topology_files(links, paths_new, base_filename=f"Dragonfly_{hosts}_topology{topo}", bandwidth=2)
    write_ns3_topology_file(links, paths_new, filename=f"ns3_Dragonfly_{hosts}_topology{topo}", bandwidth="2MB/s")
```
