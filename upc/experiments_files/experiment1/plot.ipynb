```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data from the table
data = {
    'Simulator/Protocol': [
        'Analytical', 
        'G2', 
        'ns-3 with DCTCP', 
        'ns-3 with DCQCN',
        'ns-3 with HPCC', 

    ],
    'Completion Time (s)': [0.0054, 0.1024, 0.1037, 0.1091, 0.1227]
}

df = pd.DataFrame(data)

# Create a color palette
palette = {
    'Analytical': 'salmon',
    'G2': 'skyblue',
    'ns-3 with DCTCP': 'lightgray',
    'ns-3 with HPCC': 'lightgray',
    'ns-3 with DCQCN': 'lightgray'
}

# Plotting - Vertical
plt.figure(figsize=(6, 4))
ax_v = sns.barplot(
    x='Simulator/Protocol', 
    y='Completion Time (s)', 
    data=df,
    palette=[palette[i] for i in df['Simulator/Protocol']]
)

# Add labels on the bars
for i in ax_v.containers:
    ax_v.bar_label(i, fmt='%.4f', padding=3, fontsize=12)

# Set plot labels and title
plt.xlabel('', fontsize=14)
plt.ylabel('Completion Time (s)', fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.ylim(0, df['Completion Time (s)'].max() * 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
sns.despine()
plt.show()


# Plotting - Horizontal
plt.figure(figsize=(6, 4))
ax_h = sns.barplot(
    x='Completion Time (s)', 
    y='Simulator/Protocol', 
    data=df,
    palette=[palette[i] for i in df['Simulator/Protocol']],
    orient='h'
)

# Add labels on the bars
for i in ax_h.containers:
    ax_h.bar_label(i, fmt='%.4f', padding=5, fontsize=12)

# Set plot labels and title
plt.xlabel('Completion Time (s)', fontsize=14)
plt.ylabel('', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
sns.despine()
plt.show()
```
