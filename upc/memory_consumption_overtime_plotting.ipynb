```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Load the CSV file
file_path = './output/GPT_3_1300M/2D_Torus/2_1_32_1_1_memory.csv'
data = pd.read_csv(file_path)

# Filter rows where the 'Sys' column value is 32
filtered_data = data[data[' Sys'] == 32]

# Calculate total memory consumption (sum of all memory types)
activation = (filtered_data['  Activation'] )/ 1024 / 1024  # Convert to MB
gradient = (filtered_data[' Gradient'] )/ 1024 / 1024  # Convert to MB
parameter = [max( filtered_data[' Parameter']/ 1024 / 1024)] * len(filtered_data[' Parameter'])  # Convert to MB
optimizer = [max (filtered_data[' Optimizer']/ 1024 / 1024)] * len(filtered_data[' Parameter'])  # Convert to MB
filtered_data['Total_Consumption'] = (activation + gradient + parameter + optimizer)  

# Create an interactive plot
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add traces for each memory type
fig.add_trace(
    go.Scatter(x=filtered_data[' Time'], y=activation, 
               name="Activation", mode='lines+markers', line=dict(color='blue')),
    secondary_y=False,
)

fig.add_trace(
    go.Scatter(x=filtered_data[' Time'], y=gradient, 
               name="Gradient", mode='lines+markers', line=dict(color='red')),
    secondary_y=False,
)

fig.add_trace(
    go.Scatter(x=filtered_data[' Time'], y=parameter, 
               name="Parameter", mode='lines+markers', line=dict(color='green')),
    secondary_y=False,
)

fig.add_trace(
    go.Scatter(x=filtered_data[' Time'], y=optimizer, 
               name="Optimizer", mode='lines+markers', line=dict(color='purple')),
    secondary_y=False,
)

fig.add_trace(
    go.Scatter(x=filtered_data[' Time'], y=filtered_data['Total_Consumption'], 
               name="Total_Consumption", mode='lines+markers', line=dict(color='black')),
    secondary_y=False,
)


# Update layout
fig.update_layout(
    title_text="Memory Consumption Over Time (System 32)",
    xaxis_title="Time",
    legend=dict(x=0, y=1.1, orientation='h'),
    height=600,
    width=1000
)

fig.update_yaxes(title_text="Individual Memory Components (bytes)", secondary_y=False)
fig.update_yaxes(title_text="Total Memory Consumption (bytes)", secondary_y=True)

fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

# Load the CSV file
file_path = './output/GPT_3_1300M_4/2D_Torus/2_1_32_1_0_memory.csv'
data = pd.read_csv(file_path)

# Filter rows where the 'Sys' column value is 32
filtered_data = data[data[' Sys'] == 32]

# Sort data by time to ensure correct ordering
filtered_data = filtered_data.sort_values(by=' Time')

# Calculate memory consumption components (convert to MB)
activation = (filtered_data['  Activation']) / 1024 / 1024 / 1024
gradient = (filtered_data[' Gradient']) / 1024 / 1024 / 1024

max_optimizer = filtered_data[' Optimizer'].max()
gradient_values = filtered_data[' Gradient'].values
first_nonzero_idx = np.argmax(gradient_values != 0)  # returns first True index

filtered_data[' Optimizer'].values[first_nonzero_idx:] = max_optimizer
optimizer = filtered_data[' Optimizer'] / 1024 / 1024 / 1024

parameter = [max(filtered_data[' Parameter'] / 1024 / 1024 / 1024)] * len(filtered_data[' Parameter'])
filtered_data['Total_Consumption'] = (activation + gradient + parameter + optimizer)

# Create sequential indices for x-axis
x_indices = list(range(len(filtered_data)))
time_values = filtered_data[' Time'].values

# Create a stacked bar chart
fig = go.Figure()


fig.add_trace(
    go.Bar(
        x=x_indices,
        y=parameter,
        name="Parameter",
        marker=dict(color='green'),
        hovertemplate='Time: %{customdata}<br>Parameter: %{y:.2f} GB<extra></extra>',
        customdata=time_values,
        width=1  # Set width to 1 to ensure bars touch each other
    )
)
# Add each memory component as a separate bar trace
fig.add_trace(
    go.Bar(
        x=x_indices,
        y=activation,
        name="Activation",
        marker=dict(color='blue'),
        hovertemplate='Time: %{customdata}<br>Activation: %{y:.2f} GB<extra></extra>',
        customdata=time_values,
        width=1  # Set width to 1 to ensure bars touch each other
    )
)

fig.add_trace(
    go.Bar(
        x=x_indices,
        y=gradient,
        name="Gradient",
        marker=dict(color='red'),
        hovertemplate='Time: %{customdata}<br>Gradient: %{y:.2f} GB<extra></extra>',
        customdata=time_values,
        width=1  # Set width to 1 to ensure bars touch each other
    )
)


fig.add_trace(
    go.Bar(
        x=x_indices,
        y=optimizer,
        name="Optimizer",
        marker=dict(color='purple'),
        hovertemplate='Time: %{customdata}<br>Optimizer: %{y:.2f} GB<extra></extra>',
        customdata=time_values,
        width=1  # Set width to 1 to ensure bars touch each other
    )
)


# Update layout for stacked bars with no gaps
fig.update_layout(
    barmode='stack',
    title_text="Memory Consumption Over Time (System 32)",
    yaxis_title="Memory Consumption (GB)",
    legend=dict(x=0, y=1.1, orientation='h'),
    height=600,
    width=1000,
    bargap=0,  # Remove gap between bars
    bargroupgap=0,  # Remove gap between bar groups
)

# Customize x-axis to show time values at certain intervals
step = max(1, len(filtered_data) // 10)  # Show approximately 10 tick marks
tick_vals = x_indices[::step]
tick_text = [f"{time_values[i]}" for i in tick_vals]

fig.update_xaxes(
    title_text="Time Steps",
    tickvals=tick_vals,
    ticktext=tick_text,
    tickangle=45
)

fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

# Load the CSV file
file_path = './output/GPT_3_1300M_4/2D_Torus/4_8_2_1_0_memory.csv'
data = pd.read_csv(file_path)

# Filter rows where the 'Sys' column value is 32
filtered_data = data[data[' Sys'] == 32]

# Sort data by time to ensure correct ordering
filtered_data = filtered_data.sort_values(by=' Time')

# Calculate memory consumption components (convert to MB)
activation = (filtered_data['  Activation']) / 1024 / 1024
gradient = (filtered_data[' Gradient']) / 1024 / 1024

# Fill in optimizer from first non-zero point
max_optimizer = filtered_data[' Optimizer'].max()
gradient_values = filtered_data[' Gradient'].values
first_nonzero_idx = np.argmax(gradient_values != 0)

filtered_data[' Optimizer'].values[first_nonzero_idx:] = max_optimizer
optimizer = filtered_data[' Optimizer'] / 1024 / 1024

# Parameter is constant over time
parameter = [max(filtered_data[' Parameter'] / 1024 / 1024)] * len(filtered_data)
filtered_data['Total_Consumption'] = (activation + gradient + parameter + optimizer)

# X-axis: time steps
x_indices = list(range(len(filtered_data)))
time_values = filtered_data[' Time'].values

# Create a stacked bar chart
fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=x_indices,
        y=parameter,
        name="Parameter",
        marker=dict(color='green'),
        hovertemplate='Time: %{customdata}<br>Parameter: %{y:.2f} MB<extra></extra>',
        customdata=time_values,
        width=1
    )
)

fig.add_trace(
    go.Bar(
        x=x_indices,
        y=activation,
        name="Activation",
        marker=dict(color='blue'),
        hovertemplate='Time: %{customdata}<br>Activation: %{y:.2f} MB<extra></extra>',
        customdata=time_values,
        width=1
    )
)

fig.add_trace(
    go.Bar(
        x=x_indices,
        y=gradient,
        name="Gradient",
        marker=dict(color='red'),
        hovertemplate='Time: %{customdata}<br>Gradient: %{y:.2f} MB<extra></extra>',
        customdata=time_values,
        width=1
    )
)

fig.add_trace(
    go.Bar(
        x=x_indices,
        y=optimizer,
        name="Optimizer",
        marker=dict(color='purple'),
        hovertemplate='Time: %{customdata}<br>Optimizer: %{y:.2f} MB<extra></extra>',
        customdata=time_values,
        width=1
    )
)

# Add scrollable range slider and zoomable x-axis
fig.update_layout(
    barmode='stack',
    title_text="Memory Consumption Over Time (System 32)",
    yaxis_title="Memory Consumption (MB)",
    legend=dict(x=0, y=1.1, orientation='h'),
    height=600,
    width=1000,
    bargap=0,
    bargroupgap=0,

    xaxis=dict(
        title="Time Steps",
        rangeslider=dict(visible=True),  # Range slider enabled
        type="linear"
    )
)

# Optional: clean and angled tick marks
step = max(1, len(filtered_data) // 10)
tick_vals = x_indices[::step]
tick_text = [f"{time_values[i]}" for i in tick_vals]

fig.update_xaxes(
    tickvals=tick_vals,
    ticktext=tick_text,
    tickangle=45
)

fig.show()
```
